```python

import os

annotation_files = []

for root, dirs, files in os.walk(repo_path):
    for file in files:
        if file.lower().endswith(('.json', '.xml', '.txt', '.yaml', '.yml')):
            annotation_files.append(os.path.join(root, file))

print("Total annotation/configuration files:",
      len(annotation_files))

for file in annotation_files:
    print(file)
```

```python

for file in annotation_files:
    print("\nFile:", os.path.basename(file))

    if file.endswith('.json'):
        print("Format: JSON")
    elif file.endswith('.xml'):
        print("Format: XML")
    elif file.endswith(('.yaml', '.yml')):
        print("Format: YAML")
    elif file.endswith('.txt'):
        print("Format: TXT")
```

```python

# STEP 15: Count annotation files by format

from collections import Counter
import os

format_counts = Counter(
    os.path.splitext(path)[1].lower()
    for path in annotation_files
)

print("Annotation/configuration file counts:")
for file_format, count in format_counts.items():
    print(f"{file_format}: {count}")

print("\nTotal files:", len(annotation_files))
```

```python

# STEP 22: Extract links and redirect destinations

from bs4 import BeautifulSoup
import os
import re

weights_path = os.path.join(repo_path, "model_weights")

for filename in ["download_v1_1.html", "download_v1_2.html"]:
    file_path = os.path.join(weights_path, filename)

    print("\n==========", filename, "==========")

    with open(file_path, "r", encoding="utf-8", errors="ignore") as f:
        html = f.read()

    soup = BeautifulSoup(html, "html.parser")

    print("\nLinks:")
    for link in soup.find_all("a", href=True):
        print(link["href"])

    print("\nForms:")
    for form in soup.find_all("form"):
        print("Action:", form.get("action"))
        print("Method:", form.get("method"))

    print("\nRedirects and URLs:")
    for url in re.findall(r'https?://[^\s"\']+', html):
        print(url)
```
