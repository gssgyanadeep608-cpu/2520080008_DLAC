```python

import os

dataset_path = "/content/drive/MyDrive/DLAC_Dataset"

print("Dataset folder exists:", os.path.exists(dataset_path))

print("\nDataset contents:")

for item in os.listdir(dataset_path):
    full_path = os.path.join(dataset_path, item)

    if os.path.isdir(full_path):
        print("📁 Folder:", item)
    else:
        print("📄 File:", item)
```

```python

for root, dirs, files in os.walk(dataset_path):
    level = root.replace(dataset_path, '').count(os.sep)

    indent = "  " * level

    print(f"{indent}📁 {os.path.basename(root)}/")

    for file in files[:10]:
        print(f"{indent}  📄 {file}")

    if level >= 3:
        dirs[:] = []
```

```python

for root, dirs, files in os.walk(repo_path):
    dirs[:] = [d for d in dirs if d != ".git"]

    for file in files:
        if file.lower().endswith((
            '.json', '.yaml', '.yml', '.txt',
            '.jpg', '.jpeg', '.png', '.xml'
        )):
            print(os.path.relpath(os.path.join(root, file), repo_path))
```

```python

import os

print("SEARCHING DATASET FILES\n")

extensions = ('.json', '.yaml', '.yml', '.txt', '.xml')

for root, dirs, files in os.walk(repo_path):
    dirs[:] = [d for d in dirs if d not in ['.git', '__pycache__']]

    for file in files:
        if file.lower().endswith(extensions):
            print(os.path.join(root, file))
```

```python

weights_path = os.path.join(repo_path, "model_weights")

if os.path.exists(weights_path):
    print("Model weights found:")
    for file in os.listdir(weights_path):
        print(file)
else:
    print("Model weights folder not found")
```

```python

import os

print("Checking configuration files...\n")

for root, dirs, files in os.walk(repo_path):
    for file in files:
        if file.lower() in [
            "data.yaml",
            "dataset.yaml",
            "classes.txt",
            "annotations.json",
            "instances_train.json"
        ]:
            print("Found:", os.path.join(root, file))
```

```python

image_extensions = ('.jpg', '.jpeg', '.png')

image_files = []

for root, dirs, files in os.walk(repo_path):
    for file in files:
        if file.lower().endswith(image_extensions):
            image_files.append(os.path.join(root, file))

print("Total images found:", len(image_files))

print("\nFirst 10 images:")
for img in image_files[:10]:
    print(img)
```
